# Data Audit

This notebook audits the raw retail inventory dataset for missing values, duplicates, date issues, and invalid numeric values.


```python
import os
import pandas as pd
import numpy as np

from IPython.display import display

candidates = [
    "retail_store_inventory.csv",
    os.path.join("..", "data", "raw", "retail_store_inventory.csv"),
    os.path.join("data", "raw", "retail_store_inventory.csv"),
]

data_path = next((path for path in candidates if os.path.exists(path)), None)

if data_path is None:
    raise FileNotFoundError("Dataset not found in expected locations")

df = pd.read_csv(data_path)

# Dataset ki first 5 rows
display(df.head())

# Rows, columns aur data types
print("Shape:", df.shape)
df.info()

# Statistical summary
display(df.describe())
# Har column mein missing values
print(df.isnull().sum())

# Total missing values
print("Total null values:", df.isnull().sum().sum())

# Percentage of missing values
print((df.isnull().mean() * 100).round(2))

df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

print(df["Date"].dtype)
print("Invalid dates:", df["Date"].isna().sum())

print("Duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()
# Negative forecast values
negative_forecasts = df[df["Demand Forecast"] < 0]

print("Negative forecasts:", len(negative_forecasts))
display(negative_forecasts.head(10))
df["Demand Forecast"] = df["Demand Forecast"].clip(lower=0)

print("Remaining negative forecasts:",
      (df["Demand Forecast"] < 0).sum())

for col in df.select_dtypes(include="object").columns:
    print(col, df[col].astype("string").str.strip().eq("").sum())

numeric_cols = df.select_dtypes(include="number").columns

print(df[numeric_cols].lt(0).sum())

print("Invalid dates:", df["Date"].isna().sum())

print("Invalid prices:", (df["Price"] <= 0).sum())
print("Negative inventory:", (df["Inventory Level"] < 0).sum())
print("Negative units sold:", (df["Units Sold"] < 0).sum())
print("Negative units ordered:", (df["Units Ordered"] < 0).sum())

print("Discount range:", df["Discount"].min(), df["Discount"].max())

print("Forecasts changed to zero:",
      (negative_forecasts.index.isin(df.index)).sum())

print("Minimum forecast after cleaning:",
      df["Demand Forecast"].min())

df.to_csv("retail_store_inventory_cleaned.csv", index=False)
```
